# Plan → Code → Check
Create a small signal, remove its mean, and check the result. The imported graph preserves notebook execution order.

```python
signal = [1.0, 2.0, 4.0, 5.0]
print(signal)
```

```python
mean = sum(signal) / len(signal)
centered = [value - mean for value in signal]
print(centered)
```

```python
assert abs(sum(centered)) < 1e-12
print('Check passed: centered signal has zero sum.')
```
